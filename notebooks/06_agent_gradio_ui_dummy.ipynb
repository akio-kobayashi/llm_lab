このノートブックは `06_agent_gradio_ui.ipynb` の **UI 部分だけ** を切り出した確認用の簡易版です。
実際の LLM やエージェントは呼ばず、ダミーの回答とログを返して UI のレイアウトだけを確認します。

```python
!pip install -q -U gradio

import gradio as gr
```

```python
def create_agent_ui(run_agent_fn):
    """
    06_agent_gradio_ui.ipynb と同じ最小構成の UI。
    """
    with gr.Blocks(title="AIエージェントの処理過程の可視化") as demo:
        gr.Markdown("# AIエージェントの処理過程の可視化")
        gr.Markdown("Executor と Critic による自己修正ループを確認します。")

        with gr.Row():
            with gr.Column(scale=2):
                query = gr.Textbox(
                    label="ユーザーの質問",
                    placeholder="例: 1から100までの素数の和を説明してください",
                    lines=3,
                )
                submit_btn = gr.Button("エージェントに依頼", variant="primary")

            with gr.Column(scale=3):
                final_answer = gr.Textbox(label="最終回答", lines=10)

        with gr.Accordion("思考プロセスの詳細ログ", open=True):
            log_display = gr.Markdown(label="ログ")

        submit_btn.click(
            fn=run_agent_fn,
            inputs=[query],
            outputs=[final_answer, log_display],
        )

    return demo
```

```python
def run_dummy_agent_for_ui(query):
    cleaned_query = query.strip() or "質問が未入力です。"

    final_answer = (
        "これはダミーの最終回答です。\\n\\n"
        f"入力された質問: {cleaned_query}\\n\\n"
        "本物のエージェントの代わりに、UI の見た目とログ表示だけを確認しています。"
    )

    steps = [
        (
            "Executor",
            "まず質問の要点を整理し、回答の下書きを作成したことにします。\\n"
            f"対象クエリ: {cleaned_query}",
        ),
        (
            "Critic",
            "下書きに曖昧さがないか点検し、説明をもう少し丁寧にするよう指摘したことにします。",
        ),
        (
            "Refined Answer",
            "批評を反映して、最終回答欄に表示する文章へ整えたことにします。",
        ),
    ]

    formatted_log = ""
    for role, observation in steps:
        formatted_log += f"### {role}\\n> {observation.replace('\\n', '\\n> ')}\\n\\n"

    return final_answer, formatted_log
```

## UI の起動

次のセルを実行すると、ダミーの `run_agent_for_ui` を使った Gradio UI が起動します。

```python
ui = create_agent_ui(run_dummy_agent_for_ui)
ui.launch(share=True, debug=True)
```
